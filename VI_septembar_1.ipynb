### Handling outliers, limiting classes

````python
dinos_selected10 = dinos[dinos["name"].map(name_counts) >= 10]
dinos_selected20 = dinos[dinos["name"].map(name_counts) >= 20]
dinos_selected30 = dinos[dinos["name"].map(name_counts) >= 30]
dinos_selected40 = dinos[dinos["name"].map(name_counts) >= 40]
dinos_selected50 = dinos[dinos["name"].map(name_counts) >= 50]

selected_name_counts10 = dinos_selected10["name"].value_counts()
selected_name_counts20 = dinos_selected20["name"].value_counts()
selected_name_counts30 = dinos_selected30["name"].value_counts()
selected_name_counts40 = dinos_selected40["name"].value_counts()
selected_name_counts50 = dinos_selected50["name"].value_counts()


print("Dinos over 10 records: ", len(selected_name_counts10))
print("Dinos over 10 records: ", len(selected_name_counts20))
print("Dinos over 10 records: ", len(selected_name_counts30))
print("Dinos over 10 records: ", len(selected_name_counts40))
print("Dinos over 10 records: ", len(selected_name_counts50))
````
